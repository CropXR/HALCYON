Code to generate the 4 gene dataset with multiplicative interactions. Node 1 is the only one that has an error, and it is caused by the Node_2*Node_0 bit.

```python
gnw_sim = GNWSimRunner(
    settings_path=Path("../setting_files/gnw_testing_runs_settings.txt"),
    jar_path=Path("../GRNSimulator/CropXRSimulator.jar"),
)
for n_genes in [4]:
    for sparsity in [0.01]:
        generated_data_path = Path(f"../tests/generated_test_data/n_genes_{n_genes}_sparsity_{sparsity}")
        gnw_sim.run_simulation(
            n_genes=n_genes,
            network_type=1,
            sparsity=sparsity,
            num_modules=1,
            network_name="integration_test_network",
            output_path=generated_data_path,
            keep_self_interactions=False
        )
```
